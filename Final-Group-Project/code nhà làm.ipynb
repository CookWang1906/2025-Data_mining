## **Tổng quan về đề tài**

### **Lý do chọn đề tài**

Trong bối cảnh cạnh tranh gay gắt của thị trường tài chính hiện đại, các tổ chức ngân hàng không ngừng tìm kiếm giải pháp để tối ưu hóa hiệu quả các chiến dịch Marketing trực tiếp (Direct Marketing). Sản phẩm tiền gửi có kỳ hạn (Term Deposit) là một nguồn vốn quan trọng, nhưng việc tiếp cận và thuyết phục khách hàng tiềm năng là một thách thức lớn.

Thực tế cho thấy, việc triển khai các chiến dịch tiếp thị trên diện rộng thường dẫn đến **tỷ lệ chuyển đổi thấp** và **lãng phí nguồn lực** đáng kể, thể hiện qua số lượng lớn các cuộc gọi được thực hiện đến những khách hàng không có nhu cầu. Do đó, nhu cầu chuyển đổi từ phương pháp tiếp thị đại trà sang **chiến lược tiếp thị mục tiêu (Targeted Marketing)** dựa trên dữ liệu là vô cùng cấp thiết. Bằng cách tận dụng các dữ liệu lịch sử về khách hàng và chiến dịch, ngân hàng có thể xây dựng các mô hình dự đoán và phân khúc để nhận diện chính xác đối tượng cần nhắm đến.

Nghiên cứu này ra đời nhằm mục đích khai thác sức mạnh của **Học máy (Machine Learning)** và **Khai phá dữ liệu (Data Mining)** để phân tích sâu bộ dữ liệu marketing của ngân hàng. Mục tiêu là xây dựng một **mô hình dự đoán đáng tin cậy** có khả năng xác định xác suất đăng ký thành công của từng khách hàng, từ đó tạo ra lợi thế cạnh tranh thông qua việc **tối ưu hóa quy trình ra quyết định và phân bổ nguồn lực**. Nghiên cứu sẽ góp phần giảm chi phí cho các cuộc gọi lãng phí (False Positives) và tối đa hóa lợi nhuận bằng cách tập trung nguồn lực tiếp thị vào nhóm khách hàng tiềm năng cao nhất.

### **Giới thiệu về bộ dữ liệu**
Bộ dữ liệu này liên quan đến các chiến dịch tiếp thị trực tiếp (qua điện thoại) của một tổ chức tài chính - ngân hàng tại Bồ Đào Nha. Các chiến dịch thường yêu cầu nhân viên tiếp thị phải liên lạc nhiều lần với cùng một khách hàng để xác định xem sản phẩm (khoản tiền gửi có kỳ hạn của ngàn hàng) có được đăng ký hay không. Trong bài này, chúng tôi sẽ sử dụng file dữ liệu có tên là bank-additional-full.csv (41188 dòng và 21 cột).

### **Các đặc trưng (cột) bên trong dataset**

#### Dữ liệu khách hàng ngân hàng:
- **age**: tuổi (số)
- **job**: loại công việc (chuỗi)
- **marital**: tình trạng hôn nhân (chuỗi)
- **ducation**: trình độ học vấn (chuỗi)
- **default**: có tín dụng mặc định không? (chuỗi)
- **housing**: có khoản vay nhà ở không? (chuỗi)
- **loan**: có khoản vay cá nhân không? (chuỗi)

#### Liên quan đến liên hệ cuối cùng của chiến dịch hiện tại:
- **contact**: loại hình liên lạc (chuỗi)
- **month**: tháng liên hệ cuối cùng trong năm (chuỗi)
- **day_of_week**: ngày liên hệ cuối cùng trong tuần (chuỗi)
- **duration**: thời lượng liên hệ cuối cùng, tính bằng giây (số).

#### Các thuộc tính khác:
- **campaign**: số lượng liên hệ được thực hiện trong chiến dịch này cho khách hàng này (số)
- **pdays**: số ngày đã trôi qua sau khi khách hàng được liên hệ lần cuối từ một chiến dịch trước đó (số; 999 nghĩa là chưa được liên hệ)
- **previous**: số lượng liên hệ được thực hiện trước chiến dịch này cho khách hàng này (số)
- **poutcome**: kết quả của chiến dịch tiếp thị trước đó (chuỗi)

#### Các thuộc tính bối cảnh kinh tế và xã hội:
- **emp.var.rate**: tỷ lệ thay đổi việc làm - chỉ số hàng quý (số)
- **cons.price.idx**: chỉ số giá tiêu dùng - chỉ số hàng tháng (số)
- **cons.conf.idx**: chỉ số niềm tin tiêu dùng - chỉ số hàng tháng (số)
- **euribor3m**: lãi suất Euribor 3 tháng - chỉ số hàng ngày (số)
- **nr.employed**: số lượng nhân viên - chỉ số hàng quý (số)

#### Biến đầu ra (mục tiêu):
- **y**: khách hàng có đăng ký tiền gửi có kỳ hạn không? (chuỗi: 'yes', 'no')

Nguồn dữ liệu: [Bank Marketing - UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/222/bank+marketing) on UCI.

### **Câu hỏi nghiên cứu và mục tiêu phân tích**

#### **Câu hỏi nghiên cứu**
- 1. Những yếu tố nào (nhân khẩu học, kinh tế vĩ mô, lịch sử liên hệ) là quan trọng nhất ảnh hưởng đến quyết định đăng ký tiền gửi có kỳ hạn của khách hàng?
- 2. Mô hình Học máy có thể dự đoán chính xác khách hàng nào sẽ đồng ý (y = yes) với mức độ tin cậy và hiệu suất (ví dụ: F1-score) là bao nhiêu?
- 3. Độ tuổi nào có khả năng gửi tiền có kỳ hạn cao nhất và điều gì tác động tới nó?
- 4. Khách hàng có thể được phân nhóm (segment) thành các cụm đặc trưng riêng biệt dựa trên hồ sơ và hành vi của họ không?  
- 5. Cần có những chiến lược tiếp thị khác biệt nào cho từng cụm để tối ưu hóa hiệu quả chuyển đổi?

#### **Mục tiêu phân tích để làm rõ câu hỏi nghiên cứu**

1. **Tiền xử lý và phân tích khám phá dữ liệu (EDA):** Xử lý các vấn đề dữ liệu như mất cân bằng lớp, giá trị unknown và biến rò rỉ dữ liệu (`duration`) để chuẩn bị dữ liệu cho bước mô hình hóa.

2. **Xây dựng mô hình dự đoán (Classification):** Áp dụng và so sánh hiệu suất của nhiều thuật toán học máy khác nhau (Logistic Regression, Random Forest, LightGBM, XGBoost) để tìm ra mô hình tối ưu nhất trong việc nhận diện khách hàng có tiềm năng cao đăng ký sản phẩm.

3. **Xây dựng mô hình phân cụm (Clustering):** Sử dụng các mô hình phân cụm để phân khúc khách hàng, từ đó xác định chân dung rõ ràng của các nhóm tiềm năng (ví dụ: cụm có tỷ lệ thành công cao nhất), giúp đội ngũ Marketing cá nhân hóa thông điệp và kênh tiếp cận.

Kết quả của phân tích này không chỉ cung cấp một mô hình dự đoán mạnh mẽ mà còn đưa ra các **khuyến nghị chiến lược** trực tiếp, đảm bảo rằng nguồn lực tiếp thị được phân bổ hiệu quả nhất, giúp tăng **tỷ suất hoàn vốn (ROI)** cho các chiến dịch trong tương lai.
